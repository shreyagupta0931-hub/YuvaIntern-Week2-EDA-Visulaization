```python
import pandas as pd

df = pd.read_csv("hotel_bookings_cleaned.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
print("Booking Cancellation Analysis:")

print(df["is_canceled"].value_counts())

print("\nCancellation Percentage:")

cancellation_percentage = df["is_canceled"].value_counts(normalize=True) * 100
print(cancellation_percentage)
```

```python
print("Hotel-wise Cancellation Analysis:")

hotel_cancellation = pd.crosstab(
    df["hotel"],
    df["is_canceled"],
    normalize="index"
) * 100

print(hotel_cancellation)
```

```python
print("Hotel-wise Cancellation Rate:")

hotel_cancellation_rate = (
    df.groupby("hotel")["is_canceled"]
    .mean() * 100
)

print(hotel_cancellation_rate)
```

```python
print("Market Segment-wise Cancellation Analysis:")

market_cancellation = pd.crosstab(
    df["market_segment"],
    df["is_canceled"],
    normalize="index"
) * 100

print(market_cancellation)
```

```python
print("Monthly Booking Count:")

monthly_bookings = df["arrival_date_month"].value_counts()

print(monthly_bookings)
```

```python
print("Monthly Cancellation Analysis:")

monthly_cancellation = pd.crosstab(
    df["arrival_date_month"],
    df["is_canceled"],
    normalize="index"
) * 100

print(monthly_cancellation)
```

```python
print("ADR Analysis:")

print("Average ADR:", df["adr"].mean())
print("Median ADR:", df["adr"].median())
print("Minimum ADR:", df["adr"].min())
print("Maximum ADR:", df["adr"].max())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="adr", bins=50, kde=True)

plt.title("Distribution of Average Daily Rate (ADR)")
plt.xlabel("Average Daily Rate (ADR)")
plt.ylabel("Number of Bookings")

plt.show()
```

```python
print("Monthly Average ADR:")

monthly_adr = (
    df.groupby("arrival_date_month")["adr"]
    .mean()
)

print(monthly_adr)

```

```python
plt.figure(figsize=(10, 6))

monthly_adr.plot(kind="bar")

plt.title("Average Daily Rate (ADR) by Arrival Month")
plt.xlabel("Arrival Month")
plt.ylabel("Average ADR")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
print("Lead Time Analysis:")

print("Average Lead Time:", df["lead_time"].mean())
print("Median Lead Time:", df["lead_time"].median())
print("Minimum Lead Time:", df["lead_time"].min())
print("Maximum Lead Time:", df["lead_time"].max())
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="lead_time", bins=50, kde=True)

plt.title("Distribution of Booking Lead Time")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Number of Bookings")

plt.show()
```

```python
print("Lead Time vs Cancellation:")

lead_time_cancellation = (
    df.groupby("is_canceled")["lead_time"]
    .mean()
)

print(lead_time_cancellation)
```

```python
plt.figure(figsize=(8, 5))

lead_time_cancellation.plot(kind="bar")

plt.title("Average Lead Time by Booking Cancellation")
plt.xlabel("Booking Cancellation (0 = Not Cancelled, 1 = Cancelled)")
plt.ylabel("Average Lead Time (Days)")

plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```
